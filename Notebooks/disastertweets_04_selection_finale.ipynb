### Introduction
Après avoir testé plusieurs architectures allant des modèles de base (Logistic Regression, SVM) aux architectures de Deep Learning (LSTM, TextCNN) et aux modèles transformateurs de pointe (BERT, RoBERTa, DeBERTa), nous procédons ici à la comparaison finale des performances pour choisir le modèle à mettre en production.

Le choix du modèle se base principalement sur le **F2-Score** afin de minimiser les faux négatifs (tweets de catastrophe non détectés) tout en gardant une précision acceptable.

### Choix du Modèle Final
Sur la base des résultats observés :

1. **Performance brute** : Le modèle **BERTweet**  affiche les meilleurs scores de F2-score et de rappel sur la **classe 1**
2. **Efficacité** : Bien que plus lourd, le gain en performance justifie son utilisation pour cette tâche critique.
3. **Interpretabilité** : Les modèles transformateurs permettent également une analyse fine des tokens influents via ***(SHAP)***.

**Modèle choisi : BERTweet**
